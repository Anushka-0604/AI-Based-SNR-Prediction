```python

import pandas as pd
import numpy as np

# ==========================================
# STEP 1: CHECK AND HANDLE EXACT DUPLICATES
# ==========================================

print("Original sensor dataset shape:", sensors.shape)

exact_duplicates = sensors.duplicated().sum()
duplicate_ids = sensors['id'].duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)
print("Duplicate transmission IDs:", duplicate_ids)

# Remove only completely identical records
sensors = sensors.drop_duplicates().copy()

print("Shape after removing exact duplicates:", sensors.shape)


# ==========================================
# STEP 2: HOURLY AGGREGATION PER NODE
# ==========================================

# Ensure datetime columns exist
sensors['datetime'] = pd.to_datetime(
    sensors['timestamp'], unit='s'
)

sensors['hour'] = sensors['datetime'].dt.floor('h')

# Function to find most frequent categorical value
def most_frequent(series):
    modes = series.mode()
    return modes.iloc[0] if not modes.empty else np.nan

# Aggregate individual transmissions into hourly records
sensors_hourly = sensors.groupby(
    ['hour', 'nodeid']
).agg(
    transmission_count=('id', 'count'),

    battery_mean=('battery', 'mean'),
    frequency_mean=('frequency', 'mean'),

    gateway_rssi_mean=('gtw_rssi', 'mean'),
    gateway_snr_mean=('gtw_snr', 'mean'),

    soil_temp_mean=('soil_temp', 'mean'),
    soil_hum_mean=('soil_hum', 'mean'),

    data_rate=('data_rate', most_frequent),
    coding_rate=('coding_rate', most_frequent),
    gateway_channel=('gtw_channel', most_frequent)

).reset_index()


# ==========================================
# STEP 3: CHECK AGGREGATED DATASET
# ==========================================

print("\nHourly aggregated dataset shape:")
print(sensors_hourly.shape)

print("\nNumber of records per node:")
print(sensors_hourly.groupby('nodeid').size())

print("\nMissing values after aggregation:")
print(sensors_hourly.isnull().sum())

print("\nSample hourly records:")
display(sensors_hourly.head(10))
```

```python

import pandas as pd
import numpy as np

# ==========================================
# STEP 1: CONVERT COMBINED DATA TO LONG FORMAT
# ==========================================

combined_records = []

for i in range(1, 9):

    record = df[
        ['hour', 'temp', 'hum', 'bar', 'rain',
         f'RSSI_{i:02d}', f'SNR_{i:02d}']
    ].copy()

    record.columns = [
        'hour', 'temp', 'hum', 'bar', 'rain',
        'RSSI', 'SNR'
    ]

    record['nodeid'] = f'tinovi-{i:02d}'

    combined_records.append(record)

combined_long = pd.concat(
    combined_records,
    ignore_index=True
)

# ==========================================
# STEP 2: MERGE BOTH DATASETS
# ==========================================

final_df = pd.merge(
    combined_long,
    sensors_hourly,
    on=['hour', 'nodeid'],
    how='left',
    indicator=True
)

# ==========================================
# STEP 3: CHECK MERGE QUALITY
# ==========================================

print("Combined hourly records:", len(combined_long))
print("Final merged records:", len(final_df))

print("\nMerge status:")
print(final_df['_merge'].value_counts())

print("\nMissing values after merging:")
print(final_df.isnull().sum())

# Remove merge indicator
final_df.drop(columns=['_merge'], inplace=True)

# ==========================================
# STEP 4: BASIC DATASET INFORMATION
# ==========================================

print("\nFinal dataset shape:", final_df.shape)

print("\nRecords per node:")
print(final_df.groupby('nodeid').size())

print("\nSample of integrated dataset:")
display(final_df.head(10))
```

```python

# =====================================================
# AI-BASED SNR PREDICTION
# FEATURE ENGINEERING + MODEL TRAINING + EVALUATION
# =====================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.base import clone

# =====================================================
# STEP 1: SELECT INPUT FEATURES AND TARGET
# =====================================================

features = [
    'RSSI',
    'temp',
    'hum',
    'bar',
    'rain',
    'soil_temp_mean',
    'soil_hum_mean',
    'battery_mean',
    'transmission_count',
    'nodeid'
]

target = 'SNR'

X = ml_df[features].copy()
y = ml_df[target].copy()

# =====================================================
# STEP 2: CHRONOLOGICAL TRAIN-TEST SPLIT
# =====================================================

# Split by unique timestamps to prevent same-hour
# observations from entering both training and testing.

unique_hours = np.sort(ml_df['hour'].unique())

split_position = int(len(unique_hours) * 0.80)

split_time = unique_hours[split_position]

train_mask = ml_df['hour'] < split_time
test_mask = ~train_mask

X_train = X.loc[train_mask]
X_test = X.loc[test_mask]

y_train = y.loc[train_mask]
y_test = y.loc[test_mask]

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Training ends:", ml_df.loc[train_mask, 'hour'].max())
print("Testing starts:", ml_df.loc[test_mask, 'hour'].min())

# =====================================================
# STEP 3: PREPROCESSING
# =====================================================

numeric_features = [
    'RSSI', 'temp', 'hum', 'bar', 'rain',
    'soil_temp_mean', 'soil_hum_mean',
    'battery_mean', 'transmission_count'
]

categorical_features = ['nodeid']

numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', numeric_pipeline, numeric_features),
    ('cat', categorical_pipeline, categorical_features)
])

# =====================================================
# STEP 4: DEFINE MACHINE LEARNING MODELS
# =====================================================

models = {
    'Linear Regression': LinearRegression(),

    'Decision Tree': DecisionTreeRegressor(
        max_depth=8,
        random_state=42
    ),

    'Random Forest': RandomForestRegressor(
        n_estimators=100,
        max_depth=12,
        random_state=42,
        n_jobs=-1
    ),

    'Gradient Boosting': GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    )
}

# Add XGBoost if available
try:
    from xgboost import XGBRegressor

    models['XGBoost'] = XGBRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=5,
        random_state=42,
        objective='reg:squarederror',
        n_jobs=-1
    )

except ImportError:
    print("\nXGBoost is not installed. Skipping it.")

# =====================================================
# STEP 5: TRAIN AND EVALUATE ALL MODELS
# =====================================================

results = []
trained_models = {}
predictions = {}

for name, model in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        ('preprocessor', clone(preprocessor)),
        ('model', model)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append({
        'Model': name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    })

    trained_models[name] = pipeline
    predictions[name] = y_pred

# =====================================================
# STEP 6: MODEL COMPARISON
# =====================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='RMSE',
    ascending=True
).reset_index(drop=True)

print("\n========== MODEL COMPARISON ==========")
display(results_df.round(4))

# =====================================================
# STEP 7: IDENTIFY BEST MODEL
# =====================================================

best_model_name = results_df.iloc[0]['Model']

best_model = trained_models[best_model_name]

best_predictions = predictions[best_model_name]

print("\nBest Model:", best_model_name)

print("MAE:",
      mean_absolute_error(y_test, best_predictions))

print("MSE:",
      mean_squared_error(y_test, best_predictions))

print("RMSE:",
      np.sqrt(mean_squared_error(y_test, best_predictions)))

print("R2 Score:",
      r2_score(y_test, best_predictions))

# =====================================================
# STEP 8: ACTUAL VS PREDICTED SNR
# =====================================================

plt.figure(figsize=(9, 5))

plt.scatter(
    y_test,
    best_predictions,
    alpha=0.5
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle='--'
)

plt.xlabel('Actual SNR')
plt.ylabel('Predicted SNR')
plt.title('Actual vs Predicted SNR - ' + best_model_name)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# =====================================================
# STEP 9: MODEL COMPARISON GRAPH
# =====================================================

plt.figure(figsize=(9, 5))

plt.bar(
    results_df['Model'],
    results_df['RMSE']
)

plt.xlabel('Machine Learning Models')
plt.ylabel('RMSE')
plt.title('Model Performance Comparison')

plt.xticks(rotation=30)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# =====================================================
# STEP 10: SAVE RESULTS
# =====================================================

results_df.to_csv(
    '/content/drive/MyDrive/SNR_Model_Comparison.csv',
    index=False
)

print("\nModel comparison results saved successfully!")
```
